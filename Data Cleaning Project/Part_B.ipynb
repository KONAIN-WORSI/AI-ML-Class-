```python
import pandas as pd
```

# Outlier Detection

```python
df = pd.read_csv('Stage1.csv')

def remove_outlier_iqr(data, col):
    Q1 = data[col].quantile(0.25)
    Q3 = data[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    filt = data[(data[col] >= lower_bound) & (data[col] <= upper_bound)]

    return filt

df = (
    df
    .pipe(remove_outlier_iqr, col = 'Age')
    .pipe(remove_outlier_iqr, col = 'Salary')
    .pipe(remove_outlier_iqr, col = 'PerformanceScore')
)
df.info()
df.describe()
```

As we can see from the stastics of before and after outlier removal. The mean has shifted more because outlier affects the mean directly more than the median. so, The mean after the outlier removal shifted backward or downward.
