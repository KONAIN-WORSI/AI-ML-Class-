```python
import pandas as pd
```

```python
df = pd.read_csv('Stage2.csv')

df['Gender'] = df['Gender'].str.strip().replace({'Male': 'M', 'Female': 'F'})

df['Department'] = df['Department'].str.strip().str.upper().replace({
    'SALES': 'SALES',
    'FINANCE': 'FINANCE',
    'OPERATIONS': 'OPERATIONS'
})

df  = df.drop_duplicates(subset = ['Name'])

# Re-RRunning the outlier remover function
def remove_outlier_iqr(data, col):
    Q1 = data[col].quantile(0.25)
    Q3 = data[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    filt = data[(data[col] >= lower_bound) & (data[col] <= upper_bound)]

    return filt

df = (
    df
    .pipe(remove_outlier_iqr, col = 'Age')
    .pipe(remove_outlier_iqr, col = 'Salary')
)

filt = df[((df['Age'] <= 18) | (df['Age'] >= 60)) | (df['Salary'] <= 1000)]

df = df.drop(filt.index)

df.to_csv('Stage3.csv', index = False)
```
