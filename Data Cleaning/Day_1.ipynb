# data cleaning (Basics)

```python
import pandas as pd
```

```python
df  = pd.read_csv('day1_data.csv')
df
```

```python
# defining functions for data cleaning
def load_data(file_path):
    return pd.read_csv(file_path)
```

```python
def convert_numeric(df, cols):
    for col in cols:
        df[col] = pd.to_numeric(df[col], errors = 'coerce')
    return df
    
```

```python
def fill_missing(df):
    for col in df.columns:
        if df[col].dtype in ['float64' , 'int64']:
            df[col] = df[col].fillna(df[col].mean())
        else:
            df[col] = df[col].fillna(df[col].mode()[0])
    return df
```

```python
def remove_invalid_salary(df):
    return df[df['Salary'] >= 0]
```

```python
def save_result(df, file_path):
    df.to_csv(file_path, index = False)
```

```python
df = (
    load_data('day1_data.csv')
    .pipe(convert_numeric, ['Age', 'Salary'])
    .pipe(fill_missing)
    .pipe(remove_invalid_salary)
)
```

```python
save_result(df, 'Cleaned_day1_data.csv')
```
