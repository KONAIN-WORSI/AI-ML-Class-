# Data Cleaning

```python
import pandas as pd
```

```python
df = pd.read_csv('sample_dirty_dataset.csv')
df
```

```python
# remove outliers
def remove_outliers_IQR(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    return data[(data[column] >= lower_bound) & (data[column] <= upper_bound)]


df_clean = (
    df
    .pipe(remove_outliers_IQR, column = 'Age')
    .pipe(remove_outliers_IQR, column = 'Salary')
)

df_clean
# The error is that the column 'Salary' contains NaN values, which can affect the IQR calculation.
# To fix this, you should drop rows with NaN in the columns before applying the outlier removal.

# Example fix:
df_clean = (
    df.dropna(subset=['Age', 'Salary'])
      .pipe(remove_outliers_IQR, column='Age')
      .pipe(remove_outliers_IQR, column='Salary')
)

df_clean
```

```python
df_clean['Department'].fillna('Unknown', inplace  = True)

df_clean.drop_duplicates(inplace = True)

df_clean
```

```python
# Making the data consistent
df_clean['Gender'] = df_clean['Gender'].replace({'Male': 'M',
                                                 'Female': 'F'})
df_clean
```

```python
df_clean.to_csv('Cleaned_data.csv', index = False)
```
