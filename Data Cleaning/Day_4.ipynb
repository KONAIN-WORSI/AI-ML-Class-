```python
import pandas as pd
```

```python
file_path = 'raw_dataset.csv'

df = pd.read_csv(file_path)
df
```

```python
# standardize the data
df['Gender'] = df['Gender'].str.strip().str.capitalize().replace({'Male' : 'M', 'Female': 'F'})

df['Name'] = df['Name'].str.strip().str.capitalize()

df['Department'] = df['Department'].str.strip().str.capitalize().replace({
    'It' : 'IT',
    'Hr' : 'HR'
})


```

```python
# handling missing values
thres = len(df) * 0.5

df = df.dropna(thresh = thres, axis = 1)

for col in df.columns:
    if df[col].dtype in ['float64', 'int64']:
        df[col].fillna(df[col].median())
    else:
        df[col].fillna(df[col].mode()[0])


```

```python
# convert the date and time column to datetime format
df['Joining_Date'] = pd.to_datetime(df['Joining_Date'])
df['Joining_Date'] = df['Joining_Date'].fillna(df['Joining_Date'].mode()[0])

```

```python
# Ensuring correct data types
df['Age'] = df['Age'].astype(float)
df['Salary'] = df['Salary'].astype(float)
df['Performance_Score'] = df['Performance_Score'].astype(float)

```

```python
# removing outliers
def remove_outliers(df, col):
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    return df[(df[col] >= lower_bound) & (df[col] <= upper_bound)]

df = (
    df
    .pipe(remove_outliers, 'Age')
    .pipe(remove_outliers, 'Salary')
    .pipe(remove_outliers, 'Performance_Score')
)

df = df.drop_duplicates()

cleaned_path = 'cleaned_dataset.csv'

df.to_csv(cleaned_path, index = False)
df
```
