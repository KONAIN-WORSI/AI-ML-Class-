```python
import pandas as pd
```

```python
df = pd.read_csv('Stage3.csv')

df['JoiningDate'] = pd.to_datetime(df['JoiningDate'], format = '%Y-%m-%d')

thres = len(df) *  0.5

df = df.dropna(thresh = thres, axis = 1)

filt = df[(df['Age'] < 20) | (df['Salary'] > 80000)].assign(flag = '🚩')

df.loc[filt.index, 'flag'] = '🚩'

df.to_csv('Final_Cleaned.csv')
df.info()
df.describe()
```
