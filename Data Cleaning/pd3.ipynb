```python
import pandas as pd
```

```python
data = {
"A": [10 , 20, 30, 40] ,
"B": [5, 15, 25, 35] ,
"C": [2, 4, 6, 8]
}
df = pd.DataFrame(data)
df
```

Basic arithemetic operations


```python
df['A  +  B'] = df['A'] + df['B']
df['A - B'] = df['A'] - df['B']
df['A * B'] = df['A'] * df['B']
df['A / B'] = df['A'] / df['B']
df
```

Statistics

```python
df.mean()
df.median()
df.std()
df.max()
df.min()
```

Element wise operation

```python
import numpy as np

df['sq_c'] = np.sqrt(df['C'])
df['log_A'] = np.log(df['A'])
df['exp_B'] = np.exp(df['B'])
df
```

Apply


```python
# apply
def square(x):
    return x ** 2

df['Square_A'] = df['A'].apply(square)
df
```

# column wise and row wise operation


```python
# Sum across columns (row - wise )
df[" row_sum "] = df.sum ( axis =1)

# Sum across rows (column - wise )
col_sum = df.sum ( axis =0)
df
```
